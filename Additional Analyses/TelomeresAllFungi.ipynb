# Telomeres in *A. rouxii*, *M. flavus* and *M. racemosus*

```python
from Bio import SeqIO # Makes Python understand biological sequence files.
import re # Used to find patterns in text
```

## Telomeres in *A. rouxii*

```python

motif = "TTAGGG"  # adjust if needed
min_repeats = 3 # Must be atleast 3 repeats of TTAGGG

with open("/Users/emma/Documents/NMBU/Master/Master/Data/nfcore HiC pipeline files/ggAmyRoux1_genome.fasta") as f, open("telomeres_AmyRoux.bed", "w") as out:
    for record in SeqIO.parse(f, "fasta"):
        seq = str(record.seq)
        for match in re.finditer(f"({motif}){{{min_repeats},}}", seq):
            start, end = match.start(), match.end()
            out.write(f"{record.id}\t{start}\t{end}\n")
```
